```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
```

```python
import os

print("Train:", len(os.listdir("/content/dataset/train/images")))
print("Validation:", len(os.listdir("/content/dataset/valid/images")))
print("Test:", len(os.listdir("/content/dataset/test/images")))
```

```python
results = model.train(
    data="/content/dataset/data.yaml",
    epochs=30,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/drive/MyDrive/Textile_Project",
    name="textile_damage_gpu_v1",
    pretrained=True,
    cache=True
)
```

```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/Textile_Project/textile_damage_gpu_v1/weights/best.pt")

print("Model Loaded Successfully")
```

```python
results = model.predict(
    source="/content/dataset/test/images",
    conf=0.25,
    save=True,
    show_labels=True,
    show_conf=True
)
```
